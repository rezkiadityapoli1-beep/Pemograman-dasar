```python
# Latihan 3 - Cek Tipe Data PLTS
nama_unit = "PLTS Nunukan"
jumlah_panel = 240
tegangan_keluaran = 48.5
sedang_beroperasi = True

print("--- Informasi Unit PLTS ---")
print(f"Unit   : {nama_unit} ({type(nama_unit)})")
print(f"Panel  : {jumlah_panel} unit ({type(jumlah_panel)})")
print(f"Volt   : {tegangan_keluaran} V ({type(tegangan_keluaran)})")
print(f"Status : {sedang_beroperasi} ({type(sedang_beroperasi)})")
```

```python
# Latihan 4 - Pengolahan List Tegangan
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Akses elemen pertama & terakhir
print("Data awal     :", tegangan_harian[0], "V")
print("Data akhir    :", tegangan_harian[-1], "V")

# (b) Tambah data baru
tegangan_harian.append(222.0)

# (c) Hitung total elemen
print("Total sampel  :", len(tegangan_harian))
```

```python
# Latihan 5 - Dictionary Spesifikasi Motor
motor_listrik = {
    "nama": "Motor Servo DC",
    "daya_kW": 7.5,
    "tegangan_V": 220,
    "rpm": 1200
}

# Ambil nilai berdasarkan key
print("Kapasitas daya :", motor_listrik["daya_kW"], "kW")

# Update nilai rpm
motor_listrik["rpm"] = 1800

# Tampilkan seluruh dictionary terbaru
print("Data Motor Terbaru :", motor_listrik)
```

```python
 #=== LATIHAN 6 ===
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"}
]

print("Nama unit kedua:", pembangkit[1]["nama"])
pembangkit[0]["status"] = "siaga"
print("\nSetelah ubah status Unit-1:", pembangkit[0])
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 200, "status": "operasi"}
pembangkit.append(unit_baru)
print("\n=== Semua Data Pembangkit ===")
for unit in pembangkit:
    print(unit)
```
